# Exam marks: a first look

A small class, three subjects. We load the marks, find each student's **average**, and check who needs help.

- Data: a plain Python dictionary
- Tools: `pandas` for the table


```python
marks = {
    "Asha": [91, 84, 77],
    "Ben": [58, 62, 70],
    "Chen": [88, 95, 90],
    "Divya": [45, 52, 61],
}
print("Students:", len(marks))
print("Subjects: maths, science, english")
```

```python
import pandas as pd

df = pd.DataFrame(marks, index=["maths", "science", "english"]).T
df["average"] = df.mean(axis=1).round(2)
df
```

## Who needs help?

Anyone averaging under 60. A typo in the column name raises an error, which the viewer shows the way Jupyter does:

```python
df[df["avg"] < 60]
```

```python
needs_help = df[df["average"] < 60].index.tolist()
print("Needs help:", needs_help)
```
